```python
# ==============================================================================
# 🛡️ 3. Keep-Alive Daemon (Prevents Colab Idle Disconnect)
# ==============================================================================
# @title 🛡️ 3. Activate Keep-Alive Background Daemon
import threading
import time

def keep_alive():
    while True:
        time.sleep(180)

t = threading.Thread(target=keep_alive, daemon=True)
t.start()
print("✅ Keep-alive background daemon active. Colab will stay connected!")
```
